## PROJECT STATEMENT

Analyzing these factors manually from raw accident data can make it difficult to identify important patterns and understand the major contributors to accidents.

Therefore, this project uses Python-based data analysis and visualization to explore accident data, identify significant patterns, and understand the factors associated with road accidents and their severity.

<h1>DATA VISUALIZATION</h1>

```python
sns.countplot(x='Accident_severity',data=df_acc,color='green')
plt.title('Severity analysis')
plt.show()
```

```python
sns.countplot(x='Day_of_week',data=df_acc,color='blue')
plt.title('Days based accident')
plt.show()
```

```python
df_acc['Sex_of_driver'].value_counts().plot.pie(autopct='%1.1f%%')
plt.title('Gender analysis')
plt.show()
```

```python
age_group=df_acc['Age_band_of_driver'].value_counts()
age_group.plot(kind='barh',color='purple')
plt.title('Driver age group')
plt.show()
```

```python
numeric_acc=df_acc.select_dtypes(include='number')
numeric_acc
```

```python
top_causes=df_acc['Cause_of_accident'].value_counts()
top_causes.plot(kind='bar',color='Black')
plt.title('Top causes of accidents')
plt.show()
```

```python
plt.figure(figsize=(8,3))

sns.countplot(x='Weather_conditions',hue='Accident_severity', data=df_acc)
plt.title('Weather condition')
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.countplot(x='Type_of_vehicle', data=df_acc,color='Orange')
plt.xticks(rotation=90)
plt.title('Vehicle Types Involved')
plt.show()
```

```python
plt.figure(figsize=(8,5))
sns.heatmap(numeric_acc.corr(), annot=True, cmap='coolwarm')
plt.show()
```
